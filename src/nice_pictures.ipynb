```python
import pandas as pd
df = pd.read_csv("../data/examples_DBAASP/pcp_descriptors_positive_staphylococcus.csv")
```

```python
# PCP_descriptors - 5-dimensional data visualization

from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

tsne = TSNE(n_components=2, random_state=0)
df_tsne = tsne.fit_transform(df)
plt.scatter(df_tsne[:, 0], df_tsne[:, 1])

# show five graphs with gradient colors for each features
for i in range(5):
    plt.scatter(df_tsne[:, 0], df_tsne[:, 1], c=df.iloc[:, i], cmap='viridis')
    plt.colorbar()
    plt.show()
```
